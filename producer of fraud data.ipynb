```python
from confluent_kafka import Producer
import json
import time
import random
from faker import Faker
```

```python
fake = Faker()
# Kafka configuration
conf = {
    'bootstrap.servers': '', #'YOUR_BOOTSTRAP_SERVER'
    'security.protocol': '',
    'sasl.mechanisms': '',
    'sasl.username': '',             #'YOUR_API_KEY'
    'sasl.password': '', #'YOUR_API_SECRET'
}
```

```python
#CONNECTING PRODUCER WITH KAFKA
producer = Producer(conf)
def generate_transactions():
    transaction = {
        "transaction_id" : fake.uuid4(),
        "timestamp" : int(time.time()),
        "user_id": random.randint(10000, 99999),
        "amount": round(random.uniform(5,5000), 2),
        "transaction_type": random.choice(["purchase", "transfer","withdrawal"]),
        "location":fake.city(),
        "merchant":fake.company(),
        "card_number":fake.credit_card_number()
    }
    return transaction

def delivery_report(err,msg):
    if err is not None:
        print(f"message is not deliverd: {err}")
    else:
        print(f"msg deliverd to {msg.topic()},{[{msg.partition()}]},{msg}")

while True:
    txn = generate_transactions()
    #tnxdata is the topic created in the kafka
    producer.produce('txndata',value = json.dumps(txn),key=str(txn['user_id']),callback = delivery_report)
    producer.flush()
    time.sleep(1)
```
